```python
import pandas as pd

df = pd.read_csv("../data/raw/vehicles.csv")
```

# PHASE 1: Basic Missing Values/Remove Bad Columns

```python
print(df.info())
```

```python
def check_missing_rate(df):
    df_null = df.isnull().mean() * 100
    print(f"Length of dataframe: {len(df)}")
    print(df_null[df_null>0].apply(lambda x: f"{(x):.2f}%"))
    del df_null
```

```python
check_missing_rate(df)
```

## Dropping irrelevant columns, and removing rows with invalid years

```python
processed_df = df.drop(columns = ["VIN","url","region_url","county","size"])
print(f"Length of the dataset - {len(processed_df)}")

print("\nStatistical analysis of 'year' feature")
print(processed_df["year"].describe())

print("\nQuantiles in 0.05, 0.25, 0.5, 0.75 of year")
print(processed_df["year"].quantile([0.05, 0.25, 0.5, 0.75]))

processed_df = processed_df[processed_df["year"] > 1900]
print(f"Length of the dataset after dropping invalid years - {len(processed_df)}")
```

## Analyze price, removing invalid prices

```python
print("Statistical analysis of price")
pd.options.display.float_format = '{:.2f}'.format
print(processed_df["price"].describe())
pd.reset_option('display.float_format')

print("\nQuantiles in 0.05, 0.25, 0.5, 0.75, 0.999 of price")
print(processed_df["price"].quantile([0.05, 0.25, 0.5, 0.75, 0.999]))

above_120k_cars_df = processed_df[processed_df["price"] > processed_df["price"].quantile(0.999)]

print("\nManufacturer analysis of cars above the 0.999 quantile")
print(above_120k_cars_df["manufacturer"].value_counts().nlargest(10))
print(len(above_120k_cars_df))

print("Value analysis of car prices above 0.999 quantile")
pd.options.display.float_format = '{:.2f}'.format
print(above_120k_cars_df["price"].quantile([0.05, 0.25, 0.5, 0.75, 0.999]))

processed_df.nlargest(
    100,
    "price"
)[[
    "price",
    "manufacturer",
    "model",
    "year",
]]

processed_df = processed_df[
    (processed_df["price"] > 0) &
    (processed_df["price"] <= 700000)
]
print(f"Length of the dataset after dropping invalid prices - {len(processed_df)}")
```

## Analyze odometer

```python
import numpy as np

print("Statistical analysis of odometer")
pd.options.display.float_format = '{:.2f}'.format
print(processed_df["odometer"].describe())
pd.reset_option('display.float_format')
print(f"Number of missing odometer rows: {processed_df["odometer"].isna().sum()}")

print("\nQuantiles in 0.05, 0.25, 0.5, 0.75, 0.999 of price")
print(processed_df["odometer"].quantile([0.05, 0.25, 0.5, 0.75, 0.999]))

above_112k_odometer_df = processed_df[processed_df["odometer"] > processed_df["odometer"].quantile(0.999)]

print("\nValue analysis of car odometers above 0.999 quantile")
pd.options.display.float_format = '{:.2f}'.format
print(above_112k_odometer_df["odometer"].quantile([0.05, 0.25, 0.5, 0.75, 0.999]))

print("\nLooking into odometer of 0")
odometer_0_df = processed_df[processed_df["odometer"] == 0]
print(f"Number of rows with 0 in odometer: {len(odometer_0_df)}")
print((odometer_0_df[["title_status"]]).value_counts())

print("\nLooking into year distribution with odometer of 0")
print(odometer_0_df["year"].describe())
print(
    odometer_0_df["year"]
    .value_counts()
    .sort_index()
)

processed_df = processed_df[
    (processed_df["price"] > 0) &
    (processed_df["price"] <= 700000)
]

processed_df.loc[(processed_df["odometer"] == 0) & (processed_df["odometer"] > 2322706), "odometer"] = np.nan
```

## Change posting_date to datetime

```python
print(processed_df["posting_date"].head(10))
print(type(processed_df["posting_date"].iloc[0]))
processed_df["posting_date"] = pd.to_datetime(processed_df["posting_date"])
```

## Dropping invalid lat long

```python
print(processed_df["lat"].nunique())
print(processed_df["long"].nunique())

print(len(processed_df))
perfect_match = processed_df['lat'].isna().equals(processed_df['long'].isna())
print(f"Missing lats and lons are on the same rows: {perfect_match}")

processed_df = processed_df[
    (
        processed_df["lat"].isna() |
        processed_df["lat"].between(-90, 90)
    ) &
    (
        processed_df["long"].between(-180, 180)
    )
]
check_missing_rate(processed_df)
```

# PHASE 2: Rule-based processing
- Normalize manufacturer, model
- Regex/Keyword extraction
- Feature engineering

```python
def normalize_feature(df, column):
    before = df[column].nunique(dropna=False)

    mask = df[column].notna()

    df.loc[mask, column] = (
        df.loc[mask, column]
        .astype(str)
        .str.lower()
        .str.strip()
        .str.replace(r"\s{2,}", " ", regex=True)
    )

    after = df[column].nunique(dropna=False)

    print(f"{column}: {before} -> {after}")
```

```python
print("\nRandom 20 samples of model and manufacturer")
print(processed_df[["model","manufacturer"]].sample(20))
print(f"\nUnique counts of model: {processed_df["model"].nunique()}")

print("\nValue counts of manufacturer")
print(processed_df["manufacturer"].value_counts())
print(f"Unique counts of manufacturer: {processed_df["manufacturer"].nunique()}")
print("Random 30 model samples of manufacturers 'rover' and 'land rover'")
print(processed_df.loc[
    processed_df["manufacturer"].isin(["rover", "land rover"]),
    ["manufacturer", "model"]
].sample(30, random_state=42))
print("'rover' and 'land rover' value counts")
print(processed_df["manufacturer"].value_counts().loc[
    ["rover", "land rover"]
])
```

```python
processed_df["manufacturer"] = processed_df["manufacturer"].replace(
    {"rover": "land rover"}
)

for col in [
    "manufacturer",
    "model",
    "condition",
    "cylinders",
    "fuel",
    "title_status",
    "transmission",
    "drive",
    "type",
    "paint_color",
]:
    normalize_feature(processed_df, col)

check_missing_rate(processed_df)
```

```python
for col in ["condition", "cylinders","fuel","title_status","transmission","drive","type","paint_color"]:
    print(f"\n{processed_df[col].value_counts()}")
```
